## python中使用ollama

```python
import ollama

# 聊天式
response = ollama.chat(model="deepseek-r1:8b", messages=[{"role": "user", "content": "七八个星天外"}])
print(response)
print(response["message"])
```

```python
## 远程调用的方式
from ollama import Client

client = Client(host="http://127.0.0.1:11434") # 虚拟机的需要改ip
response = client.chat(model="deepseek-r1:8b", messages=[{"role": "user", "content": "完全两个东西..."}])
print(response["message"]['content'])
```

```python
## 流式调用
stream = client.chat(model="deepseek-r1:8b", messages=[{"role": "user", "content": "乘舟将欲行"}], stream=True)
for chunk in stream:
    print(chunk["message"]['content'], end="")
```

## langchain调用

```python
from langchain_community.llms import Ollama  # ✅ 你环境里有，直接能用

# from langchain_community.llms import Ollama

llm = Ollama(base_url="http://127.0.0.1:11434", model="deepseek-r1:8b")
llm.invoke("而我独缺")
```

## HTTP调用

```python
# 发送POST
# POST GET
import requests

url = "http://127.0.0.1:11434/api/chat"
headers = {"Content-Type": "application/json"} # 请求头
data = {"model": "deepseek-r1:8b", "messages": [{"role": "user", "content": "僧敲月下门"}], "stream": False}
response = requests.post(url, headers=headers, json=data)
print(response.json())
```

```python
url = "http://127.0.0.1:11434/api/chat"
headers = {"Content-Type": "application/json"}
data = {"model": "deepseek-r1:8b", "messages": [{"role": "user", "content": "亲亲小鹅鹅"}], "stream": True}
response = requests.post(url, headers=headers, json=data)

for chunk in response.iter_lines():
    print(chunk.decode("utf-8"))
```

## openai sdk调用Ollama

```python
# !pip install openai
from openai import OpenAI

client = OpenAI(base_url="http://localhost:11434/v1", api_key="xx") # vllm部署可以指定api_key

stream = client.chat.completions.create(
    model="deepseek-r1:8b",
    messages=[{"role": "user", "content": "写一首关于秋天的五言诗"}],
    stream=True,
)
# vllm
for chunk in stream:
    if chunk.choices and chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)

print()  # 换行
```
